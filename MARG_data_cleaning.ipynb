```python
import re
from datetime import datetime

def extract_report_date(filename):
    name = filename.lower()

    months = {
        "january": 1, "february": 2, "march": 3,
        "april": 4, "may": 5, "june": 6,
        "july": 7, "august": 8, "september": 9,
        "october": 10, "november": 11, "december": 12
    }

    for month, num in months.items():
        if month in name:
            year = re.search(r"(20\d{2})", name)
            if year:
                return pd.Timestamp(int(year.group()), num, 1)

    return pd.NaT

df["snapshot_date"] = df["report_file"].apply(extract_report_date)

print(df["snapshot_date"].isna().sum())
print(df["snapshot_date"].min())
print(df["snapshot_date"].max())
```

```python
import re
import numpy as np

def normalize_project_name(name):
    if pd.isna(name):
        return np.nan

    name = str(name).upper().strip()
    name = re.sub(r"\s+", " ", name)

    return name

# Normalize project names
df["project_name_normalized"] = df["project_name"].apply(
    normalize_project_name
)

# Create internal project ID
df["project_id"] = np.where(
    df["project_code"].notna(),
    df["project_code"].astype(str) + "_" +
    df["project_name_normalized"].fillna("UNKNOWN"),
    "NAME_" + df["project_name_normalized"].fillna("UNKNOWN")
)

# Check duplicates within each monthly snapshot
dup_project = (
    df.groupby(["snapshot_date", "project_id"])
      .size()
      .reset_index(name="count")
)

print(
    "Project ID/snapshot duplicates:",
    (dup_project["count"] > 1).sum()
)

print(
    dup_project[dup_project["count"] > 1]
    .sort_values("count", ascending=False)
    .head(20)
)
```

```python
#FEATURE ENGINEERING
df = df.sort_values(
    ["project_id", "snapshot_date"]
).reset_index(drop=True)
```

```python
# Previous snapshot values
df["prev_progress"] = df.groupby("project_id")[
    "physical_progress_pct"
].shift(1)

df["prev_expenditure"] = df.groupby("project_id")[
    "cumulative_expenditure_cr"
].shift(1)

df["prev_revised_cost"] = df.groupby("project_id")[
    "revised_cost_cr"
].shift(1)
```

```python
# Progress growth
df["progress_growth"] = (
    df["physical_progress_pct"] -
    df["prev_progress"]
)

# Expenditure growth
df["expenditure_growth"] = (
    df["cumulative_expenditure_cr"] -
    df["prev_expenditure"]
)

# Revised cost growth
df["cost_growth"] = (
    df["revised_cost_cr"] -
    df["prev_revised_cost"]
)
```

```python
df["expenditure_vs_original_ratio"] = (
    df["cumulative_expenditure_cr"] /
    df["original_cost_cr"]
)

df["expenditure_vs_revised_ratio"] = (
    df["cumulative_expenditure_cr"] /
    df["revised_cost_cr"]
)
```

```python
print(
    df[
        [
            "project_id",
            "snapshot_date",
            "original_cost_cr",
            "revised_cost_cr",
            "cumulative_expenditure_cr",
            "expenditure_vs_original_ratio",
            "expenditure_vs_revised_ratio"
        ]
    ].head(20)
)
```

```python
df["months_since_prev_snapshot"] = (
    df.groupby("project_id")["snapshot_date"]
      .diff()
      .dt.days / 30.44
)

df[[
    "project_id",
    "snapshot_date",
    "months_since_prev_snapshot"
]].head(20)
```

```python
df["progress_velocity"] = (
    df["progress_growth"] / df["months_since_prev_snapshot"]
)

df["expenditure_velocity"] = (
    df["expenditure_growth"] / df["months_since_prev_snapshot"]
)

df[[
    "project_id",
    "snapshot_date",
    "progress_growth",
    "expenditure_growth",
    "months_since_prev_snapshot",
    "progress_velocity",
    "expenditure_velocity"
]].head(20)
```

```python
df["cost_overrun_ratio"] = (
    (df["revised_cost_cr"] - df["original_cost_cr"])
    / df["original_cost_cr"]
)

df["cost_overrun_cr"] = (
    df["revised_cost_cr"] - df["original_cost_cr"]
)

df[[
    "project_id",
    "snapshot_date",
    "original_cost_cr",
    "revised_cost_cr",
    "cost_overrun_cr",
    "cost_overrun_ratio"
]].head(20)
```

```python
df["planned_duration_days"] = (
    df["revised_doc"] - df["revised_start_date"]
).dt.days

df["elapsed_duration_days"] = (
    df["snapshot_date"] - df["revised_start_date"]
).dt.days

df[[
    "project_id",
    "snapshot_date",
    "revised_start_date",
    "revised_doc",
    "planned_duration_days",
    "elapsed_duration_days"
]].head(20)
```

```python
df["planned_end_date"] = df["revised_doc"].fillna(df["target_doc"])

df["planned_duration_days"] = (
    df["planned_end_date"] - df["revised_start_date"]
).dt.days

df["elapsed_duration_days"] = (
    df["snapshot_date"] - df["revised_start_date"]
).dt.days

df[[
    "project_id",
    "snapshot_date",
    "revised_start_date",
    "target_doc",
    "revised_doc",
    "planned_end_date",
    "planned_duration_days",
    "elapsed_duration_days"
]].head(20)
```

```python
df["time_elapsed_ratio"] = np.where(
    df["planned_duration_days"] > 0,
    df["elapsed_duration_days"] / df["planned_duration_days"],
    np.nan
)

df["time_elapsed_ratio"] = df["time_elapsed_ratio"].clip(lower=0)

df["expected_progress_pct"] = (
    df["time_elapsed_ratio"] * 100
).clip(0, 100)

df["progress_gap"] = (
    df["physical_progress_pct"] - df["expected_progress_pct"]
)

df["schedule_pressure"] = (
    df["expected_progress_pct"] - df["physical_progress_pct"]
)

df[[
    "project_id",
    "snapshot_date",
    "physical_progress_pct",
    "time_elapsed_ratio",
    "expected_progress_pct",
    "progress_gap",
    "schedule_pressure"
]].head(20)
```

```python
df["remaining_progress_pct"] = (
    100 - df["physical_progress_pct"]
)

df["expenditure_intensity"] = np.where(
    df["physical_progress_pct"] > 0,
    df["expenditure_vs_original_ratio"] /
    (df["physical_progress_pct"] / 100),
    np.nan
)

df[[
    "project_id",
    "snapshot_date",
    "physical_progress_pct",
    "remaining_progress_pct",
    "expenditure_vs_original_ratio",
    "expenditure_intensity"
]].head(20)
```

```python
# Check the engineered features
feature_cols = [
    "prev_progress",
    "prev_expenditure",
    "prev_revised_cost",
    "progress_growth",
    "expenditure_growth",
    "cost_growth",
    "expenditure_vs_original_ratio",
    "expenditure_vs_revised_ratio",
    "months_since_prev_snapshot",
    "progress_velocity",
    "expenditure_velocity",
    "cost_overrun_cr",
    "cost_overrun_ratio",
    "planned_duration_days",
    "elapsed_duration_days",
    "time_elapsed_ratio",
    "expected_progress_pct",
    "progress_gap",
    "schedule_pressure",
    "remaining_progress_pct",
    "expenditure_intensity"
]

print("Number of engineered features:", len(feature_cols))
print("\nMissing values:")
print(df[feature_cols].isna().sum())

print("\nInfinite values:")
print(np.isinf(df[feature_cols].select_dtypes(include=np.number)).sum())

print("\nFinal shape:", df.shape)
```
